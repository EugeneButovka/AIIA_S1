### Example 1: Circuit with an X-gate (NOT)
This circuit implements the equivalent of a classical NOT.

```python
# Set the intial state of the simulator to the ground state using from_int
state = Statevector.from_int(0, 2) # |0>

# Create circuit with a X (NOT) gate
qc = QuantumCircuit(1) # takes 1 qubit
qc.x(0) # apply X to qubit 0

# Print the circuit
print(qc) # qc.draw('mlp') outside Colab to get nicer circuits

# Evolve the state by the quantum circuit
state = state.evolve(qc)

# Obtain probabilities
print(state.probabilities_dict())

# Print state in Latex style
state.draw('latex')
```

### Example 2: Circuit with an H-gate (Hadamard)
In this circuit we define a qubit in a superposition state with 50% probability for each state using the Hadamard gate.

```python
# Set the intial state of the simulator to the ground state using from_int
state = Statevector.from_int(0, 2) # |0>

# Create circuit with a Hadamard gate (superposition of qubit)
qc = QuantumCircuit(1) # takes 1 qubit
qc.h(0) # apply H to qubit 0

# Print the circuit
print(qc) # qc.draw('mlp') outside Colab to get nicer circuits

# Evolve the state by the quantum circuit
state = state.evolve(qc)

# Obtain probabilities
print(state.probabilities_dict())

# Print state in Latex style
state.draw('latex')
```

### Example 3: Circuit with controlled gate
In this example we can see how to use a bit as a control of another, specifically for a gate X, i.e., a classical CNOT gate. It can also be used for the rest of quantum gates by putting a 'c' in front (see [documentation](https://qiskit.org/documentation/stubs/qiskit.circuit.library.CXGate.html)).

```python
# Set the intial state of the simulator to the ground state using from_int
state = Statevector.from_int(2, 2**2) # |10>

# Create circuit with a CNOT gate
qc = QuantumCircuit(2) # takes 2 qubits
# Apply controlled gate: note that qiskit uses little endian notation, a higher index implies more relevance
qc.cx(1,0)

# Print the circuit
print(qc) # qc.draw('mlp') outside Colab to get nicer circuits

# Evolve the state by the quantum circuit
state = state.evolve(qc)

# Obtain probabilities
print(state.probabilities_dict())

# Print state in Latex style
state.draw('latex')
```

### Exercise 1
Entanglement is when two or more qubits have bound properties. For example in the state  $\frac{\sqrt2}{2}|00\rangle + \frac{\sqrt2}{2}|11\rangle$, the value of the second qubit is determined by the value of the first.
If they are not entangled the state would be \frac{1}{2}$|00\rangle + \frac{1}{2}|01\rangle + \frac{1}{2}|10\rangle + \frac{1}{2}|11\rangle$.

Complete the code below so that the output of the two qubits is entangled. Expected state: $\frac{\sqrt2}{2}|00\rangle + \frac{\sqrt2}{2}|11\rangle$.

```python
state = Statevector.from_int(0, 2**2)
qc = QuantumCircuit(2)

#========================

qc.h(0)
qc.cx(0, 1)

#========================
state = state.evolve(qc)
print(qc)
print(state.probabilities_dict())
state.draw('latex')
```

### Exercise 4
Starting from the provided state $Ψ = \sqrt{0.5}|00\rangle + \sqrt{0.5}|11\rangle$ use the necessary gates to obtain the state $|00\rangle$ with 100% probability.

```python
state = Statevector([1/math.sqrt(2)+0.j,0.+0.j, 0.+0.j,1/math.sqrt(2)+0.j])
qc = QuantumCircuit(2)
#========================

qc.cx(0, 1)
qc.h(0)

#========================
state = state.evolve(qc)
print(qc)
print( {key: f'{value:.2f}' for key, value in state.probabilities_dict().items() if value > 0.01})
state.draw('latex')
```

### Exercise 5
Qubit 0 has 20% probability of $|0\rangle$ and 80% probability of $|1\rangle$ and qubit 1 has 100% probability of $|0\rangle$.

Swap them to get $\sqrt{0.2}|00\rangle + \sqrt{0.8}|01\rangle$.

```python
# Aplicar las amplitudes dadas a los estados |00> y |10>
state = Statevector([math.sqrt(0.2), 0, math.sqrt(0.8), 0])
qc = QuantumCircuit(2)
#========================

qc.swap(0, 1)

#========================
state = state.evolve(qc)
print(qc)
print( {key: f'{value:.2f}' for key, value in state.probabilities_dict().items() if value > 0.01})
state.draw('latex')
```

## Analysis

The exercises above demonstrate the basic concepts of quantum computing: superposition, entanglement, interference, and the reversibility of quantum gates.

**Superposition.** A Hadamard gate puts a qubit in an equal superposition of $|0\rangle$ and $|1\rangle$. In Exercises 2 and 3, two Hadamard gates keep all four combinations of Q1 and Q2 present at once, so a single Toffoli (plus two CNOTs) evaluates a classical AND/OR function over every input simultaneously — the essence of quantum parallelism.

**Entanglement.** Exercise 1 builds the Bell state $\frac{\sqrt{2}}{2}|00\rangle + \frac{\sqrt{2}}{2}|11\rangle$: the two qubits can no longer be described independently, and measuring one instantly determines the other. Exercise 4 shows the reverse — since every gate is unitary, undoing the entangling circuit (CX then H) returns the system to $|00\rangle$ with 100% probability.

**Amplitudes vs. probabilities.** In Exercise 6 the weight of $|10\rangle$ and $|11\rangle$ is consolidated into $|11\rangle$: probabilities add ($0.4 + 0.4 = 0.8$) but amplitudes do not ($\sqrt{0.4} + \sqrt{0.4} \ne \sqrt{0.8}$). The controlled-H makes the two branches interfere constructively on $|0\rangle$ and destructively on $|1\rangle$, and a final CX moves the result to $|1\rangle$. Quantum evolution acts on amplitudes, and interference is what turns them into probabilities.

**Phases matter.** Exercise 7 works only because of the phase of $|-\rangle$, which is invisible in probabilities. Since $X|+\rangle = |+\rangle$, the order of H and X is what maps $|+\rangle \rightarrow |1\rangle$ and $|1\rangle \rightarrow |-\rangle$, allowing the closing controlled-Hadamard to reconstruct the basis state $|11\rangle$.

**Swapping qubit states.** Exercises 5 and 8 use SWAP to exchange the states of two qubits (internally three CNOTs): the whole amplitude distribution moves between the qubits, exchanging the probabilities of $|01\rangle$ and $|10\rangle$ while leaving $|00\rangle$ and $|11\rangle$ untouched.

**Reversible logic.** Classical AND/OR destroy information and are not reversible, but Exercises 2 and 3 show that with an ancilla target qubit they become unitary operations. OR is realized through the identity $a \lor b = a \oplus b \oplus (a \wedge b)$ — two CNOTs compute the XOR and a Toffoli adds the AND correction.
